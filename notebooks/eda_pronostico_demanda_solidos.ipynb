El modelo actual debe tratarse como baseline. La siguiente fase deberia comparar alternativas por horizonte y granularidad. Para empezar, tiene sentido probar modelos sobre combinaciones SOLIDO de alto volumen, como carnation / white, carnation / red y minicarnation / light pink.
